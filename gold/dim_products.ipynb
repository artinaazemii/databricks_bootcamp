```python
%sql
USE CATALOG workspace;
USE SCHEMA gold;
```

#Transformation Logic

```python
df_dim_products = spark.sql("""
SELECT DISTINCT
    product_id,
    product_key,
    product_name,
    product_line,
    product_cost
FROM silver.crm_products
""")
```

```python
df_dim_products.display()
```

#Writing Gold Table

```python
df_dim_products.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold.dim_products")
```

```python
%sql
select * from workspace.gold.dim_products
```

#Sanity Checks

```python
%sql
SELECT product_id, COUNT(*)
FROM workspace.gold.dim_products
GROUP BY product_id
HAVING COUNT(*) > 1;
```

```python
%sql
COMMENT ON TABLE workspace.gold.dim_products IS
'Product dimension table containing product attributes such as name, product line, and cost. Used to analyze sales by product.';
```
