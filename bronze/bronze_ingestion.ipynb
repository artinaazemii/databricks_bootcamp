#Ingested all CSV files into Bronze tables.

```python
crm_base_path = "/Volumes/workspace/bronze/raw_sources/source_crm/"
erp_base_path = "/Volumes/workspace/bronze/raw_sources/source_erp/"

sources = {
    # CRM
    crm_base_path + "cust_info.csv": "bronze.crm_cust_info",
    crm_base_path + "prd_info.csv": "bronze.crm_prd_info",
    crm_base_path + "sales_details.csv": "bronze.crm_sales_details",

    # ERP
    erp_base_path + "CUST_AZ12.csv": "bronze.erp_cust_az12",
    erp_base_path + "LOC_A101.csv": "bronze.erp_loc_a101",
    erp_base_path + "PX_CAT_G1V2.csv": "bronze.erp_px_cat_g1v2"
}

for file_path, table_name in sources.items():
    print(f"Loading {file_path} → {table_name}")

    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(file_path)
    )

    (
        df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(table_name)
    )

    print(f"Loaded {table_name}")
```
